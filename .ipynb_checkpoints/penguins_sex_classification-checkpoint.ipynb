# Palmer Penguins Sex Classification — Midterm Project

## 1. Problem Framing
* **Problem Statement:** Predict the biological sex (`male` vs. `female`) of Palmer Archipelago penguins based on physical body measurements (`culmen_length_mm`, `culmen_depth_mm`, `flipper_length_mm`, `body_mass_g`) and habitat attributes (`species`, `island`).
* **Why It Matters:** Non-invasive sex determination in seabirds helps ecologists and conservationists monitor population demographics and breeding ratios without stressful or costly genetic testing.
* **Evaluation Metric:** **ROC AUC** (Receiver Operating Characteristic Area Under the Curve), well-suited for binary classification tasks.

## 2. Imports & Configuration

```python
import pickle
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.feature_extraction import DictVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split, KFold
from sklearn.metrics import mutual_info_score, roc_auc_score
from sklearn.tree import DecisionTreeClassifier, export_text
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier

warnings.filterwarnings("ignore")
```

## 3. Data Preparation & Cleaning
* **Data Source:** Palmer Penguins dataset (`/workspaces/Midterm-Project/data/penguins_size.csv`).

```python
data = "/workspaces/Midterm-Project/data/penguins_size.csv"
df = pd.read_csv(data)
df.head()
```

```python
df.isnull().sum()
```

```python
df.nunique()
```

```python
df.sex.unique()
```

Replace the invalid `'.'` entry in the `sex` column with `NaN`, inspect rows with missing values, and drop rows where `sex` is missing (those rows also contain missing morphological features).

```python
df['sex'] = df['sex'].replace(['.'], np.nan)
df[df["sex"].isna()]
```

```python
df = df.dropna(subset=['sex']).reset_index(drop=True)
df.isnull().sum()
```

```python
# Standardize column names and string values
df.columns = df.columns.str.replace(' ', '_').str.lower()

categorical_columns = list(df.dtypes[df.dtypes == 'object'].index) + list(df.dtypes[df.dtypes == 'str'].index)
for c in set(categorical_columns):
    df[c] = df[c].str.replace(' ', '_').str.lower()

# Encode target variable: 1 = male, 0 = female
df.sex = (df.sex == "male").astype(int)
df.head()
```

## 4. Exploratory Data Analysis (EDA)
Explore target balance, feature distributions, outliers, mutual information, and correlations.

```python
numerical = ["culmen_length_mm", "culmen_depth_mm", "flipper_length_mm", "body_mass_g"]
categorical = ["species", "island"]

# 1. Target class balance
print("Target distribution (1 = Male, 0 = Female):")
print(df['sex'].value_counts(normalize=True))

# 2. Summary statistics & outlier check
df[numerical].describe()
```

```python
# 3. Distributions across species and sex
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, col in zip(axes.flatten(), numerical):
    sns.boxplot(data=df, x='species', y=col, hue='sex', ax=ax)
    ax.set_title(f'{col} by Species & Sex')
plt.tight_layout()
plt.show()
```

```python
# 4. Feature correlation matrix
plt.figure(figsize=(8, 6))
sns.heatmap(df[numerical].corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Numerical Feature Correlation Matrix")
plt.show()
```

## 5. Validation Framework (Train / Validation / Test Split)
Split the cleaned dataset into **60% train**, **20% validation**, and **20% test** sets using `random_state=1` for reproducibility.

```python
df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=1)

df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
df_full_train = df_full_train.reset_index(drop=True)

y_train = df_train.sex.values
y_val = df_val.sex.values
y_test = df_test.sex.values

del df_train['sex']
del df_val['sex']
del df_test['sex']

len(df_train), len(df_val), len(df_test), len(df_full_train)
```

```python
# Feature importance via Mutual Information on full_train
def mutual_info(series):
    return mutual_info_score(series, df_full_train.sex)

info = df_full_train[numerical + categorical].apply(mutual_info)
info.sort_values(ascending=False)
```

```python
# Prepare one-hot encoded feature matrices for train and validation sets
dv = DictVectorizer(sparse=False)
X_train = dv.fit_transform(df_train[categorical + numerical].to_dict(orient='records'))
X_val = dv.transform(df_val[categorical + numerical].to_dict(orient='records'))
```

## 6. Modeling & Hyperparameter Tuning
We train and tune **4 distinct model families**:
1. **Logistic Regression** (Baseline linear model)
2. **Decision Tree Classifier**
3. **Random Forest Classifier**
4. **Gradient Boosting Classifier**

### 6.1 Model 1: Logistic Regression (Baseline & Regularization Tuning)

```python
def train_logreg(df_tr, y_tr, C=1.0):
    dicts = df_tr[categorical + numerical].to_dict(orient='records')
    dv_lr = DictVectorizer(sparse=False)
    X_tr = dv_lr.fit_transform(dicts)

    model = LogisticRegression(C=C, max_iter=1000)
    model.fit(X_tr, y_tr)
    return dv_lr, model

def predict_model(df_va, dv_model, model):
    dicts = df_va[categorical + numerical].to_dict(orient='records')
    X_va = dv_model.transform(dicts)
    return model.predict_proba(X_va)[:, 1]
```

```python
# Tune regularization parameter C using 5-Fold Cross-Validation
n_splits = 5
kfold = KFold(n_splits=n_splits, shuffle=True, random_state=1)

for C in [0.001, 0.01, 0.1, 0.5, 1.0, 5.0, 10.0]:
    scores = []
    for train_idx, val_idx in kfold.split(df_full_train):
        df_tr_fold = df_full_train.iloc[train_idx]
        df_va_fold = df_full_train.iloc[val_idx]

        y_tr_fold = df_tr_fold.sex.values
        y_va_fold = df_va_fold.sex.values

        dv_fold, model_fold = train_logreg(df_tr_fold, y_tr_fold, C=C)
        y_pred_fold = predict_model(df_va_fold, dv_fold, model_fold)

        scores.append(roc_auc_score(y_va_fold, y_pred_fold))

    print('C=%6s -> %.4f +- %.4f' % (C, np.mean(scores), np.std(scores)))
```

### 6.2 Model 2: Decision Tree Classifier

```python
# Baseline unpruned Decision Tree
dt_base = DecisionTreeClassifier(random_state=1)
dt_base.fit(X_train, y_train)
y_pred_dt_base = dt_base.predict_proba(X_val)[:, 1]
print(f"Baseline Decision Tree Validation AUC: {roc_auc_score(y_val, y_pred_dt_base):.4f}")
```

```python
# Tune max_depth
depths = [1, 2, 3, 4, 5, 6, 10, 15, 20, None]

for depth in depths:
    dt = DecisionTreeClassifier(max_depth=depth, random_state=1)
    dt.fit(X_train, y_train)
    y_pred = dt.predict_proba(X_val)[:, 1]
    auc = roc_auc_score(y_val, y_pred)
    print('%4s -> %.4f' % (depth, auc))
```

```python
# Tune max_depth and min_samples_leaf together
scores_leaves = []

for depth in [3, 4, 5, 6]:
    for s in [1, 2, 5, 10, 15, 20, 30, 50]:
        dt = DecisionTreeClassifier(max_depth=depth, min_samples_leaf=s, random_state=1)
        dt.fit(X_train, y_train)
        y_pred = dt.predict_proba(X_val)[:, 1]
        auc = roc_auc_score(y_val, y_pred)
        scores_leaves.append((depth, s, auc))

df_scores = pd.DataFrame(scores_leaves, columns=['max_depth', 'min_samples_leaf', 'auc'])
df_scores_pivot = df_scores.pivot(index='min_samples_leaf', columns=['max_depth'], values=['auc'])

plt.figure(figsize=(8, 5))
sns.heatmap(df_scores_pivot, annot=True, fmt=".3f", cmap="YlGnBu")
plt.title("Decision Tree Validation AUC (max_depth vs. min_samples_leaf)")
plt.show()
```

```python
# Fit best Decision Tree and inspect decision rules
dt_best = DecisionTreeClassifier(max_depth=4, min_samples_leaf=10, random_state=1)
dt_best.fit(X_train, y_train)
print(f"Tuned Decision Tree Validation AUC: {roc_auc_score(y_val, dt_best.predict_proba(X_val)[:, 1]):.4f}\n")
print(export_text(dt_best, feature_names=list(dv.get_feature_names_out())))
```

### 6.3 Model 3: Random Forest Classifier

```python
rf_scores = []

for d in [3, 5, 7, 10, None]:
    for n in range(10, 201, 20):
        rf = RandomForestClassifier(
            n_estimators=n,
            max_depth=d,
            random_state=1,
            n_jobs=-1
        )
        rf.fit(X_train, y_train)
        y_pred_rf = rf.predict_proba(X_val)[:, 1]
        auc = roc_auc_score(y_val, y_pred_rf)
        rf_scores.append((d, n, auc))

df_rf_scores = pd.DataFrame(rf_scores, columns=['max_depth', 'n_estimators', 'auc'])

plt.figure(figsize=(9, 5))
for d in [3, 5, 7, 10, None]:
    df_subset = df_rf_scores[df_rf_scores.max_depth == d] if d is not None else df_rf_scores[df_rf_scores.max_depth.isna()]
    plt.plot(df_subset.n_estimators, df_subset.auc, label=f'max_depth={d}')

plt.xlabel('n_estimators')
plt.ylabel('Validation AUC')
plt.title('Random Forest Tuning: n_estimators & max_depth')
plt.legend()
plt.show()
```

```python
# Tune min_samples_leaf for Random Forest
rf_leaf_scores = []

for s in [1, 2, 3, 5, 10]:
    for n in range(20, 201, 20):
        rf = RandomForestClassifier(
            n_estimators=n,
            max_depth=5,
            min_samples_leaf=s,
            random_state=1,
            n_jobs=-1
        )
        rf.fit(X_train, y_train)
        y_pred_rf = rf.predict_proba(X_val)[:, 1]
        auc = roc_auc_score(y_val, y_pred_rf)
        rf_leaf_scores.append((s, n, auc))

df_rf_leaf = pd.DataFrame(rf_leaf_scores, columns=['min_samples_leaf', 'n_estimators', 'auc'])
df_rf_leaf.sort_values(by='auc', ascending=False).head(10)
```

### 6.4 Model 4: Gradient Boosting Classifier

```python
gb_scores = []

for lr in [0.01, 0.05, 0.1, 0.2]:
    for depth in [2, 3, 4]:
        for n in [50, 100, 150, 200]:
            gb = GradientBoostingClassifier(
                learning_rate=lr,
                max_depth=depth,
                n_estimators=n,
                random_state=1
            )
            gb.fit(X_train, y_train)
            y_pred_gb = gb.predict_proba(X_val)[:, 1]
            auc = roc_auc_score(y_val, y_pred_gb)
            gb_scores.append((lr, depth, n, auc))

df_gb_scores = pd.DataFrame(gb_scores, columns=['learning_rate', 'max_depth', 'n_estimators', 'auc'])
df_gb_scores.sort_values(by='auc', ascending=False).head(10)
```

## 7. Cross-Validation Model Comparison Summary
Compare all four tuned models across 5-Fold Cross-Validation on `df_full_train` before selecting the final model.

```python
def evaluate_model_cv(model_fn, df_full, n_splits=5):
    kfold = KFold(n_splits=n_splits, shuffle=True, random_state=1)
    scores = []

    for train_idx, val_idx in kfold.split(df_full):
        df_tr = df_full.iloc[train_idx]
        df_va = df_full.iloc[val_idx]

        y_tr = df_tr.sex.values
        y_va = df_va.sex.values

        dv_cv = DictVectorizer(sparse=False)
        X_tr = dv_cv.fit_transform(df_tr[categorical + numerical].to_dict(orient='records'))
        X_va = dv_cv.transform(df_va[categorical + numerical].to_dict(orient='records'))

        model = model_fn()
        model.fit(X_tr, y_tr)

        y_pred = model.predict_proba(X_va)[:, 1]
        scores.append(roc_auc_score(y_va, y_pred))

    return np.mean(scores), np.std(scores)

models = {
    "Logistic Regression (C=1.0)": lambda: LogisticRegression(C=1.0, max_iter=1000),
    "Decision Tree (max_depth=4, min_samples_leaf=10)": lambda: DecisionTreeClassifier(max_depth=4, min_samples_leaf=10, random_state=1),
    "Random Forest (n_estimators=100, max_depth=5, min_samples_leaf=2)": lambda: RandomForestClassifier(n_estimators=100, max_depth=5, min_samples_leaf=2, random_state=1),
    "Gradient Boosting (lr=0.1, max_depth=3, n_estimators=100)": lambda: GradientBoostingClassifier(learning_rate=0.1, max_depth=3, n_estimators=100, random_state=1),
}

summary = []
for name, fn in models.items():
    mean_auc, std_auc = evaluate_model_cv(fn, df_full_train)
    summary.append({
        "Model": name,
        "CV Mean ROC AUC": round(mean_auc, 4),
        "CV Std": round(std_auc, 4)
    })

df_summary = pd.DataFrame(summary).sort_values(by="CV Mean ROC AUC", ascending=False).reset_index(drop=True)
df_summary
```

## 8. Final Model Training, Test Evaluation & Model Export
Train the final selected model on the entire `df_full_train` dataset, evaluate it once on the held-out `df_test` set, and export `(dv, model)` to `model.bin` for deployment.

```python
dv_final = DictVectorizer(sparse=False)
X_full_train = dv_final.fit_transform(df_full_train[categorical + numerical].to_dict(orient='records'))
y_full_train = df_full_train.sex.values

X_test = dv_final.transform(df_test[categorical + numerical].to_dict(orient='records'))

# Select best performing model from CV comparison (update if Random Forest / Gradient Boosting scores higher)
final_model = LogisticRegression(C=1.0, max_iter=1000)
final_model.fit(X_full_train, y_full_train)

y_pred_test = final_model.predict_proba(X_test)[:, 1]
test_auc = roc_auc_score(y_test, y_pred_test)
print(f"Final Test Set ROC AUC: {test_auc:.4f}")
```

```python
output_file = "model.bin"
with open(output_file, "wb") as f_out:
    pickle.dump((dv_final, final_model), f_out)

print(f"Saved DictVectorizer and final model to {output_file}")
```
